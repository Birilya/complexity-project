# Выводы

1. **Оценка $(2 - \frac{2}{k})$ не нарушена.**  
   На всех протестированных графах отношение `EFF/OPT` не превысило теоретическую верхнюю границу.

2. **Точность на практике:**

| Класс графов | Среднее EFF/OPT | Худшее `EFF/OPT` |
|-------------|----------------|----------------|
| Все графы | 1.0 – 1.14 | 1.33 (k=3) |
| Связные графы | 1.0 – 1.10 | 1.33 |
| По числу рёбер | 1.0 – 1.11 | 1.33 |
| Деревья | **всегда 1.0** | **всегда 1.0** |

   На деревьях EFFICIENT находит **минимальный** k-разрез, потому что в дереве каждый разрез — удаление одного ребра, и жадный алгоритм совпадает с оптимальным.

3. **Граница $(2 - \frac{2}{k})$ сильно завышена для реальных графов.**  
   Даже худшее отношение 1.33 значительно ниже теоретического *max_bound* = $2 - \frac{2}{k}$. Графы, на которых bound достигается, — искусственные конструкции (см [SV]).

4. **Пик ошибки — при $(k \approx n/2)$.**  

5. **EFFICIENT всегда точен при $k=2$ и $k=n$.**  
   Для $k=2$ дерево Гомори-Ху находит глобальный минимальный разрез. Для k=n разрез единственен — удаление всех рёбер.

6. **Выигрыш по времени.**  
   EFFICIENT для n $\le$ 16 работает за десятки микросекунд (несмотря на ассимпотику $O(|V|^2|E|^2)$, из графиков видно, что алгоритм работает почти линейно по $E$). Оптимальный алгоритм уже при $n=9$ требует нескольких минут.
